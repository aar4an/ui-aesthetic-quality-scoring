```python
import os
import shutil
import pandas as pd

DATASET_DIR = r'C:\Projects\Internship_Learning\dataset'
IMAGES_DIR = os.path.join(DATASET_DIR, 'images')
LABELS_CSV = os.path.join(DATASET_DIR, 'labels.csv')
NEW_IMAGES_BASE = os.path.join(DATASET_DIR, 'new_images')

# Purani labels.csv padhna (score, remarks sahit)
df_old = pd.read_csv(LABELS_CSV)
records = df_old.to_dict('records')
existing_filenames = set(df_old['filename'])

label_to_meta = {
    'good': {'score': 8, 'remarks': 'clean layout, good spacing, strong visual hierarchy'},
    'average': {'score': 5, 'remarks': 'acceptable design but lacks refinement'},
    'poor': {'score': 2, 'remarks': 'cluttered layout, low contrast, weak hierarchy'}
}

moved_count = 0
for folder_name in ['good', 'average', 'poor']:
    src_folder = os.path.join(NEW_IMAGES_BASE, folder_name)
    if not os.path.exists(src_folder):
        print(f"Warning: {src_folder} exist nahi karta, skip kar raha hoon")
        continue
    for fname in sorted(os.listdir(src_folder)):
        src_path = os.path.join(src_folder, fname)
        dst_path = os.path.join(IMAGES_DIR, fname)
        if os.path.exists(dst_path) or fname in existing_filenames:
            name, ext = os.path.splitext(fname)
            fname = f"{name}_new{ext}"
            dst_path = os.path.join(IMAGES_DIR, fname)
        shutil.move(src_path, dst_path)
        records.append({
            'filename': fname,
            'label': folder_name,
            'score': label_to_meta[folder_name]['score'],
            'remarks': label_to_meta[folder_name]['remarks']
        })
        moved_count += 1

df = pd.DataFrame(records).sort_values('filename').reset_index(drop=True)
df.to_csv(LABELS_CSV, index=False)

print(f"Naye images move hui: {moved_count}")
print(f"Total rows ab: {len(df)}")
print("\nLabel distribution:")
print(df['label'].value_counts())
```

```python
# Dataset ko 90% train aur 10% temporary (validation + test) me split karta hai
# stratify se label distribution (good/average/poor) train aur temp dono me maintain rehta hai
train_df, temp_df = train_test_split(df, test_size=0.1, random_state=42, stratify=df['label'])

# Temporary dataset ko validation aur test me 50-50 split karta hai
# NOTE: temp_df bahut chhota hai (~5 images), isliye stratify hata di —
# kisi class ke paas itni kam images hain ki stratify se split possible nahi tha
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)

# Total images ki sankhya print karta hai
print("Total images:", len(df))
# Train set me total images print karta hai
print("\nTrain set:", len(train_df))
# Validation set me total images print karta hai
print("Validation set:", len(val_df))
# Test set me total images print karta hai
print("Test set:", len(test_df))

# Train set ka label distribution print karta hai
print("\nTrain label distribution:")
print(train_df['label'].value_counts())
# Validation set ka label distribution print karta hai
print("\nValidation label distribution:")
print(val_df['label'].value_counts())
# Test set ka label distribution print karta hai
print("\nTest label distribution:")
print(test_df['label'].value_counts())
```

```python
# Train dataset ko train.csv file me save karta hai
train_df.to_csv(os.path.join(DATASET_DIR, 'train.csv'), index=False)

# Validation dataset ko val.csv file me save karta hai
val_df.to_csv(os.path.join(DATASET_DIR, 'val.csv'), index=False)

# Test dataset ko test.csv file me save karta hai
test_df.to_csv(os.path.join(DATASET_DIR, 'test.csv'), index=False)

# Train CSV save hone ke baad total rows print karta hai
print("train.csv saved:", len(train_df), "rows")

# Validation CSV save hone ke baad total rows print karta hai
print("val.csv saved:", len(val_df), "rows")

# Test CSV save hone ke baad total rows print karta hai
print("test.csv saved:", len(test_df), "rows")
```

```python
import os
from collections import Counter
from PIL import Image

root = r"C:\Projects\Internship_Learning"
exts = (".png", ".jpg", ".jpeg", ".webp")

sizes = Counter()
examples = {}
for dirpath, _, files in os.walk(root):
    if any(skip in dirpath for skip in (".git", "venv", "site-packages", ".ipynb_checkpoints")):
        continue
    for f in files:
        if f.lower().endswith(exts):
            p = os.path.join(dirpath, f)
            try:
                with Image.open(p) as im:
                    sizes[im.size] += 1
                    examples.setdefault(im.size, dirpath)
            except Exception:
                pass

for size, n in sizes.most_common(10):
    print(size, n, "images, e.g. in", examples[size])
```
