```python
# OpenCV library import karta hai (image processing ke liye)
import cv2

# NumPy library import karta hai (arrays aur numerical operations ke liye)
import numpy as np

# Pandas library import karta hai (CSV aur tabular data handle karne ke liye)
import pandas as pd

# File aur folder operations ke liye os module import karta hai
import os

# PIL library se Image module import karta hai (image open aur manipulate karne ke liye)
from PIL import Image

# Confirm karta hai ki sabhi libraries successfully load ho gayi hain
print("All libraries loaded successfully!")
```

```python
# Image se features extract karne ke liye function define karta hai
def extract_features(image_path):

    # Image read karta hai
    img = cv2.imread(image_path)

    # Image ko BGR se RGB format me convert karta hai
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Image ko 224x224 size me resize karta hai
    img_resized = cv2.resize(img_rgb, (224, 224))

    # Image ko grayscale me convert karta hai
    gray = cv2.cvtColor(img_resized, cv2.COLOR_RGB2GRAY)

    # Image ko HSV color format me convert karta hai
    # HSV ka use saturation nikalne ke liye kiya jata hai
    hsv = cv2.cvtColor(img_resized, cv2.COLOR_RGB2HSV)

    # np.mean() ka use image ki average brightness calculate karne ke liye kiya jata hai
    brightness = round(float(np.mean(gray)), 2)

    # np.std() ka use image ka overall contrast calculate karne ke liye kiya jata hai
    contrast = round(float(np.std(gray)), 2)

    # Canny Edge Detection apply karta hai- learn how canny eddge algorithm detection works
    edges = cv2.Canny(gray, 50, 150)

    # Edge pixels ka percentage (edge density) calculate karta hai
    edge_density = round(float(np.sum(edges > 0) / edges.size * 100), 2)

    # Image ko pixels ki list me convert karta hai
    pixels = img_resized.reshape(-1, 3)

    # Pixels ko float32 format me convert karta hai
    pixels = np.float32(pixels)

    # K-Means clustering ke stopping criteria set karta hai
    criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 10, 1.0)

    # K-Means algorithm se dominant colors detect karta hai
    _, labels, _ = cv2.kmeans(pixels, 5, None, criteria, 3, cv2.KMEANS_RANDOM_CENTERS)

    # Unique dominant colors ki total count nikalta hai
    dominant_color_count = len(np.unique(labels))

    # Bright (white) pixels ka percentage calculate karta hai
    whitespace_ratio = round(float(np.sum(gray > 200) / gray.size * 100), 2)

    # Laplacian filter apply karke image ki sharpness calculate karta hai
    laplacian = cv2.Laplacian(gray, cv2.CV_64F)
    sharpness = round(float(laplacian.var()), 2)

    # HSV image se average saturation calculate karta hai
    saturation_mean = round(float(np.mean(hsv[:, :, 1])), 2)

    # Dark pixels ka percentage calculate karta hai
    dark_area_ratio = round(float(np.sum(gray < 50) / gray.size * 100), 2)

    # Light pixels ka percentage calculate karta hai
    light_area_ratio = round(float(np.sum(gray > 200) / gray.size * 100), 2)

    # Edge density aur dominant colors ke basis par clutter score calculate karta hai (clarify)
    clutter_score = round((edge_density * 0.5) + (dominant_color_count * 0.5), 2)

    # Sabhi extracted features dictionary ke form me return karta hai
    return {
        'brightness': brightness,
        'contrast': contrast,
        'edge_density': edge_density,
        'dominant_color_count': dominant_color_count,
        'whitespace_ratio': whitespace_ratio,
        'sharpness': sharpness,
        'saturation_mean': saturation_mean,
        'dark_area_ratio': dark_area_ratio,
        'light_area_ratio': light_area_ratio,
        'clutter_score': clutter_score
    }

# Confirm karta hai ki feature extraction function successfully define ho gaya hai
print("Feature extraction function defined!")
```

```python
# Images folder ka path define karta hai (naye local structure ke hisaab se)
IMAGES_DIR = r"C:\Projects\Internship_Learning\dataset\images"

# Images folder ki sabhi image files ki list leta hai
# Sirf valid image extensions wali files leta hai (folders/checkpoints ignore ho jate hain)
images = [f for f in os.listdir(IMAGES_DIR) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# Pehli image ka complete path banata hai
test_image = os.path.join(IMAGES_DIR, images[0])

# Pehli image se sabhi features extract karta hai
features = extract_features(test_image)

# Image ka naam print karta hai
print("Image:", images[0])

# Separator line print karta hai
print("-"*40)

# Har feature aur uski value ek-ek karke print karta hai
for feature, value in features.items():
    print(f"{feature:<25}: {value}")
```

```python
# Sabhi images ke extracted features store karne ke liye empty list banata hai
all_features = []

# Sabhi images ko ek-ek karke process karta hai
for img_name in images:
    
    # Current image ka complete path banata hai
    img_path = os.path.join(IMAGES_DIR, img_name)
    
    # Current image se features extract karta hai
    features = extract_features(img_path)
    
    # Dictionary me image ka filename add karta hai
    features['filename'] = img_name
    
    # Features ko list me store karta hai
    all_features.append(features)
    
    # Current image successfully process hone ka message print karta hai
    print(f"Processed: {img_name}")

# Sabhi images process hone ka message print karta hai (count dynamic hai, hardcoded nahi)
print(f"\nAll {len(images)} images processed!")
```

```python
# Extracted features ki list se DataFrame banata hai
df_features = pd.DataFrame(all_features)

# 'filename' column ko sabse pehle rakhne ke liye columns ki nayi list banata hai
cols = ['filename'] + [col for col in df_features.columns if col != 'filename']

# Columns ko naye order me arrange karta hai
df_features = df_features[cols]

# DataFrame ki pehli 5 rows display karta hai
print(df_features.head())

# Total rows print karta hai
print("\nTotal rows:", len(df_features))

# Total columns print karta hai
print("Total columns:", len(df_features.columns))

# features.csv file save karne ka path define karta hai (naye local structure ke hisaab se)
csv_path = r"C:\Projects\Internship_Learning\dataset\features.csv"

# DataFrame ko CSV file me save karta hai
# index=False se row numbers save nahi hote
df_features.to_csv(csv_path, index=False)

# CSV successfully save hone ka message print karta hai
print("\nfeatures.csv saved successfully!")
```

```python
import pytesseract
from PIL import Image
import os
import pandas as pd

pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

def extract_text_features(image_path):
    """
    Image se text-related features nikalta hai:
    - word_count: total detected words
    - text_coverage_ratio: text ne image ka kitna % area cover kiya
    - avg_text_x, avg_text_y: text ki average position (normalized 0-1)
    """
    img = Image.open(image_path)
    img_width, img_height = img.size
    
    # Pytesseract se detailed data nikalna (har word ka bounding box + confidence)
    data = pytesseract.image_to_data(img, output_type=pytesseract.Output.DICT)
    
    total_text_area = 0
    word_count = 0
    sum_x = 0
    sum_y = 0
    
    # Har detected word ke through loop chalana
    for i in range(len(data['text'])):
        word = data['text'][i].strip()
        conf = int(data['conf'][i])
        
        # Sirf woh words consider karo jinka confidence 0 se zyada hai aur khaali nahi hain
        if word != "" and conf > 0:
            word_count += 1
            
            w = data['width'][i]
            h = data['height'][i]
            x = data['left'][i]
            y = data['top'][i]
            
            total_text_area += w * h
            
            # Word ka center point add karna (average nikalne ke liye)
            sum_x += (x + w / 2)
            sum_y += (y + h / 2)
    
    # Text coverage ratio: total text area / total image area
    image_area = img_width * img_height
    text_coverage_ratio = total_text_area / image_area if image_area > 0 else 0
    
    # Average x, y position (normalized 0-1, taaki alag image sizes compare ho sakein)
    avg_text_x = (sum_x / word_count / img_width) if word_count > 0 else 0
    avg_text_y = (sum_y / word_count / img_height) if word_count > 0 else 0
    
    return {
        'word_count': word_count,
        'text_coverage_ratio': round(text_coverage_ratio, 4),
        'avg_text_x': round(avg_text_x, 4),
        'avg_text_y': round(avg_text_y, 4)
    }

# Test ek image pe
test_result = extract_text_features(test_image_path)
print("Test image ke text features:")
print(test_result)
```

```python
DATASET_DIR = r'C:\Projects\Internship_Learning\dataset'
# Saari images ke liye text features nikalna
text_features_list = []

image_files = sorted(os.listdir(IMAGES_DIR))
total = len(image_files)

for idx, filename in enumerate(image_files):
    image_path = os.path.join(IMAGES_DIR, filename)
    
    try:
        features = extract_text_features(image_path)
        features['filename'] = filename
        text_features_list.append(features)
    except Exception as e:
        # Agar koi image corrupt ho ya process na ho paye, error print karke skip karo
        print(f"Error processing {filename}: {e}")
    
    # Progress dikhane ke liye (har 20 images ke baad)
    if (idx + 1) % 20 == 0:
        print(f"Processed {idx + 1}/{total} images")

# List ko DataFrame mein convert karna
text_features_df = pd.DataFrame(text_features_list)

# Column order thoda clean rakhna (filename pehle)
text_features_df = text_features_df[['filename', 'word_count', 'text_coverage_ratio', 'avg_text_x', 'avg_text_y']]

print("\nTotal images processed:", len(text_features_df))
print("\nSample rows:")
print(text_features_df.head())

# CSV mein save karna
output_path = os.path.join(DATASET_DIR, 'text_features.csv')
text_features_df.to_csv(output_path, index=False)
print(f"\nSaved to: {output_path}")
```

```python
# Text features ko labels.csv ke saath merge karke dekhte hain
# ki different quality categories mein text patterns kaise differ karte hain

labels_df = pd.read_csv(os.path.join(DATASET_DIR, 'labels.csv'))
merged_df = pd.merge(text_features_df, labels_df[['filename', 'label']], on='filename')

print("Average text features by label:")
print(merged_df.groupby('label')[['word_count', 'text_coverage_ratio', 'avg_text_x', 'avg_text_y']].mean())
```
