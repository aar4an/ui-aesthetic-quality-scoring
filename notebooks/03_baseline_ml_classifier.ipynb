```python
# Pandas library import karta hai (tabular data aur CSV handle karne ke liye)
import pandas as pd

# NumPy library import karta hai (arrays aur numerical operations ke liye)
import numpy as np

# Logistic Regression model import karta hai (classification ke liye)
from sklearn.linear_model import LogisticRegression

# Random Forest model import karta hai (classification ke liye)
from sklearn.ensemble import RandomForestClassifier

# Support Vector Machine (SVM) model import karta hai (classification ke liye)
from sklearn.svm import SVC

# Model evaluation ke liye required metrics import karta hai
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report

# Warning messages handle karne ke liye warnings module import karta hai
import warnings

# Unnecessary warning messages ko hide karta hai
warnings.filterwarnings('ignore')

# Confirm karta hai ki sabhi libraries successfully load ho gayi hain
print("All libraries loaded!")
```

```python
# features.csv file ka path define karta hai (naye local structure ke hisaab se)
features_path = r"C:\Projects\Internship_Learning\dataset\features.csv"

# labels.csv file ka path define karta hai (naye local structure ke hisaab se)
labels_path = r"C:\Projects\Internship_Learning\dataset\labels.csv"

# Features CSV file ko DataFrame me load karta hai
df_features = pd.read_csv(features_path)

# Labels CSV file ko DataFrame me load karta hai
df_labels = pd.read_csv(labels_path)

# Features DataFrame ki shape (rows, columns) print karta hai
print("Features shape:", df_features.shape)

# Labels DataFrame ki shape (rows, columns) print karta hai
print("Labels shape:", df_labels.shape)

# Features DataFrame ke sabhi column names print karta hai
print("\nFeatures columns:", list(df_features.columns))

# Label distribution ki heading print karta hai
print("\nLabel distribution:")

# Har label kitni baar present hai uska count print karta hai
print(df_labels['label'].value_counts())
```

```python
# pd.merge() ka use features aur labels ko filename ke basis par combine karne ke liye kiya jata hai
df = pd.merge(df_features, df_labels[['filename', 'label']], on='filename')

# Features DataFrame ki total rows print karta hai
print("Features rows:", len(df_features))

# Labels DataFrame ki total rows print karta hai
print("Labels rows:", len(df_labels))

# Merge hone ke baad total rows print karta hai
print("Merged rows:", len(df))

# Model ke input features ke column names ki list DYNAMICALLY banata hai
# 'filename' aur 'label' ko chhodkar baaki sabhi columns feature maane jaate hain
# Isse future mein naye features add hone par yeh list manually update nahi karni padegi
feature_cols = [col for col in df.columns if col not in ['filename', 'label']]

# Konse features use ho rahe hain, verify karne ke liye print karta hai
print("\nFeature columns used:", feature_cols)
print("Total features:", len(feature_cols))

# Input features (X) select karta hai
X = df[feature_cols]

# Target/output labels (y) select karta hai
y = df['label']

# Input features ki shape (rows, columns) print karta hai
print("\nX shape:", X.shape)

# Target labels ki shape print karta hai
print("y shape:", y.shape)

# Label distribution ki heading print karta hai
print("\nLabel distribution:")

# Har label kitni baar present hai uska count print karta hai
print(y.value_counts())
```

```python
# train_test_split function import karta hai (dataset ko train aur test set me split karne ke liye)
from sklearn.model_selection import train_test_split

# StandardScaler import karta hai (features ko scale/normalize karne ke liye)
from sklearn.preprocessing import StandardScaler

# Dataset ko train aur test set me split karta hai
# test_size=0.1 -> 10% data testing ke liye
# random_state=42 -> Har baar same split mile
# stratify=y -> Har class ka ratio train aur test dono me same rakhta hai
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y 
)

# StandardScaler ka object banata hai
scaler = StandardScaler()

# Training data se mean aur standard deviation learn karke data scale karta hai
X_train_scaled = scaler.fit_transform(X_train)

# Training data ke same scaling parameters use karke test data scale karta hai
# Sirf transform use karte hain taaki data leakage na ho
X_test_scaled = scaler.transform(X_test)

# Train set ki shape print karta hai
print("Train size:", X_train.shape)

# Test set ki shape print karta hai
print("Test size:", X_test.shape)

# Train label distribution ki heading print karta hai
print("\nTrain label distribution:")

# Train set me har label kitni baar hai uska count print karta hai
print(y_train.value_counts())

# Test label distribution ki heading print karta hai
print("\nTest label distribution:")

# Test set me har label kitni baar hai uska count print karta hai
print(y_test.value_counts())
```

```python
# train_test_split function import karta hai (dataset ko train aur test set me split karne ke liye)
from sklearn.model_selection import train_test_split

# StandardScaler import karta hai (features ko scale/normalize karne ke liye)
from sklearn.preprocessing import StandardScaler

# Dataset ko train aur test set me split karta hai
# test_size=0.3 -> 30% data testing ke liye
# random_state=42 -> Har baar same split mile
# stratify=y -> Har class ka ratio train aur test dono me same rakhta hai
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)

# StandardScaler ka object banata hai
scaler = StandardScaler()

# Training data se mean aur standard deviation learn karke data scale karta hai
X_train_scaled = scaler.fit_transform(X_train)

# Training data ke same scaling parameters use karke test data scale karta hai
# Sirf transform use karte hain taaki data leakage na ho
X_test_scaled = scaler.transform(X_test)

# Train set ki shape print karta hai
print("Train size:", X_train.shape)

# Test set ki shape print karta hai
print("Test size:", X_test.shape)

# Train label distribution ki heading print karta hai
print("\nTrain label distribution:")

# Train set me har label kitni baar hai uska count print karta hai
print(y_train.value_counts())

# Test label distribution ki heading print karta hai
print("\nTest label distribution:")

# Test set me har label kitni baar hai uska count print karta hai
print(y_test.value_counts())
```

```python
# Different machine learning models ki dictionary banata hai
models = {
    # Logistic Regression model define karta hai
    # max_iter=1000 model ko converge hone ke liye zyada iterations deta hai
    'Logistic Regression': LogisticRegression(max_iter=1000),

    # Random Forest model define karta hai
    # n_estimators=100 -> 100 decision trees use karega
    # random_state=42 -> Har baar same result mile
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),

    # SVM model define karta hai
    # kernel='rbf' -> Non-linear decision boundary ke liye RBF kernel use karta hai
    'SVM': SVC(kernel='rbf', random_state=42)
}
#experiment with other kernels of svm
# Sabhi models ke results store karne ke liye empty dictionary banata hai
results = {}

# Har model ko ek-ek karke train aur evaluate karta hai
for model_name, model in models.items():

    # Model ko training data par train karta hai
    model.fit(X_train_scaled, y_train)

    # Test data ke labels predict karta hai
    y_pred = model.predict(X_test_scaled)

    # Accuracy calculate karta hai
    accuracy = round(accuracy_score(y_test, y_pred) * 100, 2)

    # Macro F1-score calculate karta hai
    macro_f1 = round(f1_score(y_test, y_pred, average='macro') * 100, 2)

    # Model ke results dictionary me store karta hai
    results[model_name] = {
        'accuracy': accuracy,
        'macro_f1': macro_f1,
        'y_pred': y_pred
    }

    # Separator line print karta hai
    print("="*50)

    # Model ka naam print karta hai
    print(f"Model: {model_name}")

    # Accuracy print karta hai
    print(f"Accuracy  : {accuracy}%")

    # Macro F1-score print karta hai
    print(f"Macro F1  : {macro_f1}%")

    # Classification Report ki heading print karta hai
    print("\nClassification Report:")

    # Precision, Recall aur F1-score report print karta hai
    print(classification_report(y_test, y_pred))

    # Confusion Matrix ki heading print karta hai
    print("\nConfusion Matrix:")

    # Confusion Matrix print karta hai
    print(confusion_matrix(y_test, y_pred))

    # Empty line print karta hai
    print()
```

```python
# Final results summary ki heading print karta hai
print("="*50)
print("FINAL RESULTS SUMMARY")
print("="*50)

# Table ka header print karta hai
print(f"{'Model':<25} {'Accuracy':>10} {'Macro F1':>10}")

# Separator line print karta hai
print("-"*50)

# Sabhi models ke results table format me print karta hai
for model_name, result in results.items():
    print(f"{model_name:<25} {result['accuracy']:>9}% {result['macro_f1']:>9}%")

# Closing separator print karta hai
print("="*50)

# Macro F1-score ke basis par best model select karta hai
best_model = max(results, key=lambda x: results[x]['macro_f1'])

# Best model ka naam print karta hai
print(f"\nBest Model: {best_model}")
```
